# E-commerce Sales and Customer Analysis Report

```python
# Import Libraries
import pandas as pd
import numpy as np
```

```python
# Loading the datasets
try:
    products_df = pd.read_csv('products.csv')
    orders_df = pd.read_csv('orders.csv')
    customers_df = pd.read_csv('customers.csv')
    print("Data loaded successfully.")
except FileNotFoundError:
    print("One or more CSV files not found.")
    exit()
```

```python
# Merging the datasets
merged_data = pd.merge(orders_df, customers_df, on="customer_id", how="left")
final_data = pd.merge(merged_data, products_df, on="product_id", how="left")
final_data.to_csv("merged_data.csv", index=False)
```

### Find how many customers are repeating on my store.


```python
# Read the merged file
data = pd.read_csv("merged_data.csv")

# Count how many times each customer appears
customer_order_count = data['customer_id'].value_counts()

# Filter customers who ordered more than once
repeat_customers = customer_order_count[customer_order_count > 1].index

# Print the total number of repeat customers
print("Number of repeat customers:", len(repeat_customers))
```

### Find which product is top sailing product on the basis of repeated orders.


```python
# Filter data for only repeat customers
repeat_orders = data[data['customer_id'].isin(repeat_customers)]

# Find which product is most ordered by repeat customers
top_product = repeat_orders['product_id'].value_counts().idxmax()

print("Top selling product (by repeat customers):", top_product)
```

### Find which product should be discontinued as it is the least soled product on my store.


```python
# Find the product that sold the least
least_sold = data['product_id'].value_counts().idxmin()

print("Least sold product:", least_sold)
```

Find which customer should get more discount as he/she keep visiting my store more than 5 times.


```python
# Count how many times each customer ordered
order_count = data['customer_id'].value_counts()
# Find customers with more than 5 orders
loyal_customers = order_count[order_count > 5]
print("Customers who should get more discount:")
print(loyal_customers)
```

### What is my total income till end.

```python
# Calculate total income from all orders
total_income = data['price'].sum()
print("Total income from all sales:", total_income)
```

### Write me a full report on yearly sale.


```python
# Convert date column to datetime format
data['order_date'] = pd.to_datetime(data['order_date'])
# Extract year from the date
data['year'] = data['order_date'].dt.year
# Group by year and calculate total sales for each year
yearly_sales = data.groupby('year')['price'].sum().reset_index()
print("Yearly Sales Report:")
print(yearly_sales)
```

### Write me a full report on monthly sale.

```python
# Extract year and month from the date
data['year_month'] = data['order_date'].dt.to_period('M')
# Group by month and calculate total sales
monthly_sales = data.groupby('year_month')['price'].sum().reset_index()
print("Monthly Sales Report:")
print(monthly_sales)
```
